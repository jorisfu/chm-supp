```python
import plotly.express as px
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../results/sc-runs/gmmtests-synth.csv")
df
```

```python
piv = df.pivot(index="Missingness", columns="Replicate", values="n_components")
piv
```

```python
counts = piv.agg(pd.Series.value_counts, axis=1).reset_index().melt(["Missingness"], var_name="No. of GMM components", value_name="Amount of replicates").fillna(0)
counts["Missingness"] = counts["Missingness"].apply(lambda x: x*100)
counts
```

```python
fig = px.bar(
    counts,
    x="Missingness",
    y="Amount of replicates",
    color="No. of GMM components"
)

fig.update_layout(width=800, height=400)
fig.update_xaxes(tickmode='array', tickvals=counts["Missingness"], title="Missingness in %")
fig.show()
```
