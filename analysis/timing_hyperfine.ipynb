## Distance Methods Walltime Analysis

```python
import pandas as pd
import plotly.express as px
```

```python
TIMING_FILE = "../results/time/hyperfine-clusteronly-short.json"

DISTLABELS = {
    "nandist_euclidean": "nandist Euclidean",
    "dixon_pds_sqeuclidean": "PDS",
    "euclidean": "CC Euclidean",
    "eirola_esd_gmm": "ESD/GMM",
    "eirola_esd_mvn": "ESD/MVN",
    "mesquita_eed": "EED",
}

DIST_ORDER = ["CC Euclidean", "PDS", "ESD/MVN", "ESD/GMM", "EED", "nandist Euclidean"]
```

```python
import json

with open(TIMING_FILE) as f:
    data = json.load(f)

fulldata_df = pd.DataFrame(columns=["dist", "time"])

for run in data["results"]:
    times = run["times"]
    dist = DISTLABELS[run["parameters"]["dist"]]

    for t in times:
        fulldata_df.loc[-1] = [dist, t]
        fulldata_df.index = fulldata_df.index + 1
        fulldata_df = fulldata_df.sort_index()
```

```python
condensed_df = pd.DataFrame(columns=["dist", "mean", "median", "stddev"])

for run in data["results"]:
    dist = DISTLABELS[run["parameters"]["dist"]]
    mean = run["mean"]
    median = run["median"]
    stddev = run["stddev"]

    condensed_df.loc[-1] = [dist, mean, median, stddev]
    condensed_df.index = condensed_df.index + 1
    condensed_df = condensed_df.sort_index()

condensed_df
```

```python
fig = px.box(
    fulldata_df,
    x="dist",
    y="time",
    title="ClusteredHeatMap instantiation time, clustering rows and columns, AML dataset",
    labels=DISTLABELS,
)

fig.show()
```

```python
labels={
    "dist": "Distance estimation method",
    "mean": "Mean execution time (s)",
}

fig = px.bar(
    condensed_df,
    x="dist",
    y="mean",
    error_y="stddev",
    labels=labels,
    width=500,
    height=500,
)

fig.update_xaxes(
    categoryorder="array",
    categoryarray=DIST_ORDER
)

fig.show()
```
